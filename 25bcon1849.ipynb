Factorial


```python
n = 5
fact = 1

for i in range(1, n + 1):
    fact *= i

print(f"Factorial of {n} is {fact}")
```

Fibonacci series

```python
n = 10
a = 0
b = 1

for _ in range(n):
    print(a, end=" ")
    next_term = a + b
    a = b
    b = next_term
print()
```

Dictionary

```python
student = {
    "name": "Rahul",
    "roll": 101,
    "marks": 87.5
}

student["age"] = 20

print("Name:", student["name"])
print("Roll:", student["roll"])
print("Marks:", student["marks"])
print("Age:", student["age"])
```

In C, adding a new field to a struct requires changing the struct definition. In Python, you can add a new key to a dictionary while the program is running.

Why is this useful?
It gives Python flexibility. You do not need to decide all the fields beforehand. If you later need to store an age, email, or phone number, you can add that key.

Why is it risky?
Different dictionaries can have different keys, which can make data inconsistent. A `KeyError` happens when you try to access a key that is missing; adding a new key itself does not cause an error.

The C program declares that marks is a float.The python dictionary does not declare anything.Where does the type come from?

In Python, the type comes from the value you assign, not from the dictionary itself.Python is dynamically typed: the type belongs to the object/value, and Python determines it at runtime.
